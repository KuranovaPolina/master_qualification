```python
from ultralytics import YOLO

from main import *
```

```python
YOLO_model = YOLO('model/yolo26n.pt')
DepthMap_model = DepthMap()
DistanceByClassicStereo_model = DistanceByClassicStereo()
DistanceByZoeDepth_model = DistanceByZoeDepth("ZoeD_NK")
```

```python
real_distances = []
distances_by_size = []
distances_by_classic_stereo = []
distances_by_zoe_depth = []


left_img_data_folder = '/kaggle/input/datasets/kuranovapolina/kitti-3d-object-detection-evaluation-2017/data_object_image_2/training/image_2' 
right_img_data_folder = '/kaggle/input/datasets/kuranovapolina/kitti-3d-object-detection-evaluation-2017/data_object_image_3/training/image_3' 
calib_data_folder = '/kaggle/input/datasets/kuranovapolina/kitti-3d-object-detection-evaluation-2017/data_object_calib/training/calib'
velodyne_data_folder = '/kaggle/input/datasets/kuranovapolina/kitti-3d-object-detection-evaluation-2017/data_object_velodyne/training/velodyne'

for cur_id in range(10):
    img_distances = calculate_one_image(cur_id, YOLO_model, DepthMap_model, 
                        DistanceByClassicStereo_model = DistanceByClassicStereo_model, 
                        DistanceByZoeDepth_model = DistanceByZoeDepth_model, 
                        left_img_data_folder = left_img_data_folder,
                        right_img_data_folder = right_img_data_folder,
                        calib_data_folder = calib_data_folder,
                        velodyne_data_folder = velodyne_data_folder,
                                       grid_size = 5)
    
    real_distances.extend(img_distances["real_distances"])
    distances_by_size.extend(img_distances["distances_by_size"])
    distances_by_classic_stereo.extend(img_distances["distances_by_classic_stereo"])
    distances_by_zoe_depth.extend(img_distances["distances_by_zoe_depth"])

print(real_distances)
print(distances_by_size)
print(distances_by_classic_stereo)
print(distances_by_zoe_depth)

distances_by_size_abs_rel = absRel(np.array(distances_by_size), np.array(real_distances))
distances_by_classic_stereo_abs_rel = absRel(np.array(distances_by_classic_stereo), np.array(real_distances))
distances_by_zoe_depth_abs_rel = absRel(np.array(distances_by_zoe_depth), np.array(real_distances))

print("distances_by_size_abs_rel", distances_by_size_abs_rel)
print("distances_by_classic_stereo_abs_rel", distances_by_classic_stereo_abs_rel)
print("distances_by_zoe_depth_abs_rel", distances_by_zoe_depth_abs_rel)
```
